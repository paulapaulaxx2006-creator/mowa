```python
import numpy as np
import librosa
import librosa.display
import matplotlib.pyplot as plt
```

```python
SR = 16000
```

```python
nagrania = {
    "Samogłoska": "/content/samogłoska​.m4a",
    "Wyraz": "/content/wyraz.m4a",
    "Zdanie": "/content/zdanie​.m4a"
}
```

```python
warianty_parametrow = [
    {"n_fft": 256, "hop_length": 80, "nazwa": "Wysoka rozdz. czasowa (n_fft=256, hop=80)"},
    {"n_fft": 512, "hop_length": 160, "nazwa": "Standard (n_fft=512, hop=160)"},
    {"n_fft": 2048, "hop_length": 512, "nazwa": "Wysoka rozdz. częstotliwościowa (n_fft=2048, hop=512)"}
]
```

```python
def wczytaj_audio(sciezka):
    """Wczytuje audio do 16 kHz mono."""
    y, sr = librosa.load(sciezka, sr=SR, mono=True)
    return y.astype(np.float32), sr

def generuj_reprezentacje(y, sr, n_fft, hop_length, n_mels=64):
    """Generuje log-mel spektrogram oraz MFCC dla zadanych parametrów."""
    # Log-mel spektrogram
    S = librosa.feature.melspectrogram(y=y, sr=sr, n_fft=n_fft, hop_length=hop_length, n_mels=n_mels)
    S_db = librosa.power_to_db(S, ref=np.max)

    # MFCC
    M = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=13, n_fft=n_fft, hop_length=hop_length)

    return S_db, M
```

```python
for nazwa_pliku, sciezka in nagrania.items():
    print(f"Przetwarzanie nagrania: {nazwa_pliku} ({sciezka})")
    y, sr = wczytaj_audio(sciezka)

    # Używamy standardowych parametrów z pierwszego wariantu (lub domyślnych)
    S_db, M = generuj_reprezentacje(y, sr, n_fft=512, hop_length=160)

    # Rysowanie wykresów
    fig, ax = plt.subplots(3, 1, figsize=(10, 8))

    librosa.display.waveshow(y, sr=sr, ax=ax[0])
    ax[0].set_title(f"Przebieg czasowy — {nazwa_pliku}")

    img1 = librosa.display.specshow(S_db, sr=sr, hop_length=160, x_axis="time", y_axis="mel", ax=ax[1])
    ax[1].set_title(f"Log-mel spektrogram — {nazwa_pliku}")
    fig.colorbar(img1, ax=ax[1], format="%+2.0f dB")

    img2 = librosa.display.specshow(M, sr=sr, hop_length=160, x_axis="time", ax=ax[2])
    ax[2].set_title(f"MFCC — {nazwa_pliku}")
    fig.colorbar(img2, ax=ax[2])

    plt.tight_layout()
    plt.savefig(f"analiza_{nazwa_pliku.lower()}.png") # Zapis do pliku w celu wklejenia do sprawozdania
    plt.show()
```

```python
pierwszy_klucz = list(nagrania.keys())[0]
y_test, sr_test = wczytaj_audio(nagrania[pierwszy_klucz])

for wariant in warianty_parametrow:
    print(f"Badanie wariantu: {wariant['nazwa']}")
    S_db, M = generuj_reprezentacje(y_test, sr_test, n_fft=wariant["n_fft"], hop_length=wariant["hop_length"])

    plt.figure(figsize=(10, 4))
    librosa.display.specshow(S_db, sr=sr_test, hop_length=wariant["hop_length"], x_axis="time", y_axis="mel")
    plt.colorbar(format="%+2.0f dB")
    plt.title(f"Wpływ parametrów: {wariant['nazwa']}")
    plt.tight_layout()
    plt.savefig(f"wplyw_nfft_{wariant['n_fft']}_hop_{wariant['hop_length']}.png")
    plt.show()
```
